# 06 — Pipeline D : RAG hybride

Le pipeline D reçoit le schéma récupéré et l'evidence métier de la question. Il est comparé au pipeline B pour l'hypothèse H1.

```python
from pathlib import Path
import subprocess

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
RUN_INDEX, SEED = 0, 0
RUN_NOW = False
output = f'results/study_run_{RUN_INDEX:02d}/pipeline_D/generated_sql.json'
command = [
    'python', 'src/pipeline_hybrid_rag.py',
    '--questions', 'data/processed/questions.json',
    '--index_dir', 'data/index',
    '--output', output,
    '--method', 'hybrid', '--top_k', '5',
    '--model', 'qwen2.5:7b', '--temperature', '0',
    '--seed', str(SEED), '--run_index', str(RUN_INDEX),
]
print(' '.join(command))
if RUN_NOW:
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
structural = f'results/study_run_{RUN_INDEX:02d}/pipeline_D/structural_validation.json'
execution = f'results/study_run_{RUN_INDEX:02d}/pipeline_D/execution_accuracy.json'
commands = [
    ['python', 'src/sql_validator.py', 'validate-batch', '--generated', output, '--schemas_dir', 'data/schemas', '--output', structural],
    ['python', 'src/execution_accuracy.py', '--generated', output, '--questions', 'data/processed/questions.json', '--databases_dir', 'data/raw/bird/dev_databases', '--output', execution],
]
for item in commands:
    print(' '.join(item))
    if RUN_NOW:
        subprocess.run(item, cwd=ROOT, check=True)
```
