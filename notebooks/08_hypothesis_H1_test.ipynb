# 08 — Test de l'hypothèse H1

H1 compare B (schéma) et D (schéma + connaissance métier) sur les mêmes questions tentées. Le test approprié est le **test exact de McNemar**, car les observations sont appariées.

````python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
PATH = ROOT / 'results/hypothesis_H1/h1_test_result_paired.json'
with PATH.open(encoding='utf-8') as stream:
    h1 = json.load(stream)
pd.json_normalize(h1, sep='.') .T.rename(columns={0: 'valeur'})
````

````python
def conclusion(test):
    p_value = test.get('p_value')
    if p_value is None:
        return 'Test non disponible.'
    return 'Différence statistiquement significative (p < 0,05).' if p_value < 0.05 else 'Différence non significative (p ≥ 0,05).'
tests = h1.get('tests', h1)
if isinstance(tests, dict):
    for name, test in tests.items():
        if isinstance(test, dict):
            print(name, ':', conclusion(test))
````
