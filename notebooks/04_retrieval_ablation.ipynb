# 04 — Ablations de retrieval

Comparaison des méthodes BM25, embeddings et hybride. Le choix de retrieval de schéma repose sur Table Recall@5. Le retrieval métier est rapporté séparément sur ses corpus secondaires.

````python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
RESULTS = ROOT / 'results' / 'retrieval_ablation'
def load(name):
    with (RESULTS / name).open(encoding='utf-8') as stream:
        return json.load(stream)
schema = load('schema_ablation.json')
business = load('business_ablation.json')
bird_evidence = load('bird_evidence_ablation.json')
schema.keys()
````

````python
schema_rows = schema.get('results', schema)
schema_df = pd.DataFrame(schema_rows)
display(schema_df)
metric = next((name for name in schema_df.columns if 'table' in name.lower() and 'recall' in name.lower()), None)
if metric:
    schema_df.plot.bar(x=schema_df.columns[0], y=metric, legend=False, color='#2563eb', figsize=(7, 4))
    plt.ylabel(metric); plt.title('Ablation de retrieval de schéma'); plt.tight_layout()
````

````python
for label, result in [('TPC-DS', business), ('BIRD-Evidence-Corpus', bird_evidence)]:
    rows = result.get('results', result)
    print(f'### {label}')
    display(pd.DataFrame(rows))
````
