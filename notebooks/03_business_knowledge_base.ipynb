# 03 — Base de connaissances métier

La connaissance métier utilisée par les pipelines C et D provient du champ BIRD `evidence`. Elle est injectée au niveau de chaque question ; ce n'est pas une métrique de retrieval documentaire sur BIRD.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
def load_json(relative):
    with (ROOT / relative).open(encoding='utf-8') as stream:
        return json.load(stream)
kb = load_json('data/knowledge_base/knowledge_base.json')
rules = load_json('data/business_docs/business_rules.json')
glossary = load_json('data/business_docs/glossary.json')
kpis = load_json('data/business_docs/kpis.json')
pd.Series({'knowledge_base': len(kb), 'business_rules': len(rules), 'glossary': len(glossary), 'kpis': len(kpis)}).to_frame('documents')
```

```python
pd.DataFrame(kb).head(10)
```

```python
questions = load_json('data/processed/questions.json')
evidence = pd.Series([item.get('evidence', '').strip() for item in questions])
pd.Series({
    'questions': len(questions),
    'questions avec evidence': int(evidence.ne('').sum()),
    'evidence distinctes': evidence[evidence.ne('')].nunique(),
    'longueur médiane': int(evidence.str.len().median()),
    'longueur maximale': int(evidence.str.len().max()),
}).to_frame('valeur')
```

## Interprétation

Le notebook 04 évalue le retrieval documentaire séparément sur TPC-DS et le BIRD-Evidence-Corpus. Dans l'étude principale BIRD, C et D reçoivent l'evidence liée à la question, afin de mesurer l'apport de la connaissance métier sans prétendre récupérer un document BIRD indépendant.
